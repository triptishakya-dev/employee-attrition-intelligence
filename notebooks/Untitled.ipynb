```python
## Phase 1 Findings

- Dataset contains 19,104 monthly records.
- There are 2,381 unique employees.
- Target predicts whether an employee will leave within the next 3 months.
- 23.06% of monthly observations belong to the positive attrition class.
- Employee records are longitudinal, so train/test splitting must be done by employee ID.
- Previous_Rating and Rating_Change are missing only for each employee's first observation.
- Quarterly rating, salary, tenure and business performance are being investigated as potential predictors.
- LastWorkingDate and Employee_Leaving_Date must never be used as model features because they would cause data leakage.
```

```python
X_shap = X_test_processed[:500]

if hasattr(X_shap, "toarray"):
    X_shap = X_shap.toarray()

print("SHAP sample shape:", X_shap.shape)
```

```python
shap_values = explainer.shap_values(X_shap)

print("SHAP values calculated")
```

```python
feature_names = preprocessor.get_feature_names_out()

# Newer SHAP versions may return shape:
# (samples, features, classes)
if isinstance(shap_values, np.ndarray) and shap_values.ndim == 3:
    shap_for_positive_class = shap_values[:, :, 1]
else:
    # Older SHAP versions may return a list: [class_0, class_1]
    shap_for_positive_class = (
        shap_values[1]
        if isinstance(shap_values, list)
        else shap_values
    )

print("SHAP shape:", shap_for_positive_class.shape)
print("Feature count:", len(feature_names))
```

```python
shap.summary_plot(
    shap_for_positive_class,
    X_shap,
    feature_names=feature_names
)
```

```python
# Highest-risk employee from the SHAP sample
employee_index = np.argmax(best_rf_prob[:500])

risk_score = best_rf_prob[employee_index]
actual_value = y_test.iloc[employee_index]

print("Employee sample index:", employee_index)
print("Predicted attrition risk:", round(risk_score * 100, 2), "%")
print("Actual target:", actual_value)
```

```python
feature_names = preprocessor.get_feature_names_out()

employee_shap = shap_for_positive_class[employee_index]
employee_data = X_shap[employee_index]

expected_value = explainer.expected_value

if isinstance(expected_value, (list, np.ndarray)):
    expected_value = np.array(expected_value).flatten()[-1]

employee_explanation = shap.Explanation(
    values=employee_shap,
    base_values=expected_value,
    data=employee_data,
    feature_names=feature_names
)

shap.plots.waterfall(
    employee_explanation,
    max_display=12
)
```
